#### Generators
- A generator is a convenient way, similar to writing a normal function, to construct a
new iterable object
- To create a generator, use the <b>yield</b> keyword instead of  <b>return</b> in a function:

```python
def squares(n=10):
 print(f"Generating squares from 1 to {n ** 2}")
 for i in range(1, n + 1):
     yield i ** 2
    
gen = squares()
gen
```

##### Generator expressions - gen function in single line 

```python
genexp = (x ** 2 for x in range(7))
genexp
```

```python
#### how to use generators
for x in gen: print(x) #iteration
    
#Gets next value
try:
    while True:
        print(next(gen))
except StopIteration:
    print("Done generating squares!") 
    
print(list(squares(5))) #Returns [1, 4, 9, 16, 25]

```

#### Example: Random Walks

```python
#Here is a pure Python way to implement a single random walk with 1,000 steps using
#the built-in random module:
#! blockstart
import random
position = 0
walk = [position]
nsteps = 1000
for _ in range(nsteps):
 step = 1 if random.randint(0, 1) else -1
 position += step
 walk.append(position)
#! blockend
```

```python
plt.plot(walk[:300])
```

```python
nsteps = 1000
rng = np.random.default_rng(seed=12345) # fresh random generator
draws = rng.integers(0, 2, size=nsteps) # creating array of 1 and 0 s 
steps = np.where(draws == 0, 1, -1) # replacing 0 to 1, 1 to -1 
walk = steps.cumsum() # from 0 starting cumulative sum
plt.plot(walk[:300])
```

#### Pandas Indexes

```python
idx = pd.Index(['a', 'b', 'c'])
idy = pd.Index(['c', 'd', 'e'])

## add two index objects
print(idx.append(idy))

## NOT ALL DIFFERENCEs
print(idx.difference(idy)) # which i of idx differs from idy
print(idy.difference(idx)) # which i of idy differs from idx

# common ones
print(idx.intersection(idy))

# merges vertically without repeating common ones
print(idx.union(idy))

# returns booleans: only commons will be replaced by True
print(idx.isin(idy))

# deletes elements at 2 index in Index object
print(idx.delete(2))

# Compute new Index by deleting passed values
print(idx.drop('c'))

# Compute new Index by inserting element at Index i
idx = idx.insert(2, ['f', 'f'])
print(idx) # where to insert must be specified

# Returns True if the Index has no duplicate values
print(idx.is_unique)
idx = idx.unique()
print(idx)
print(idx.is_unique)
```
