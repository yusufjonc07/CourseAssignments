## Cartesian Product

```python
import pandas as pd
import numpy as np
```

```python
table1 = pd.DataFrame([
    [10, 'a', 5],
    [15, 'b', 8],
    [25, 'a', 6],
], columns=['P', 'Q', 'R'])

table2 = pd.DataFrame([
    [10, 'b', 6],
    [15, 'c', 3],
    [10, 'b', 5],
], columns=['A', 'B', 'C'])

cp = table1.merge(table2, how='cross')
cp
```

```python
cp[cp.P==cp.A]
```

```python
cp[cp.Q==cp.B]
```

```python
cp[(cp.P==cp.A) & (cp.R==cp.C)]
```

```python
cp[(cp.P==cp.A) | (cp.R==cp.C)]
```

## Real-Table Example 0f Joins

```python
product_table = pd.read_csv('beton24_csvs.csv/hamkorqurilish_main_table_product.csv', header=None, usecols=[0, 1, 5, 6], names
                            =['Pnumber', 'PName', 'Price', 'Cno'])
product_table = product_table.iloc[:10, :]
```

```python

category_table = pd.read_csv('beton24_csvs.csv/hamkorqurilish_main_table_type_product.csv', header=None, usecols=[0, 1, 6], names
                            =['Cno', 'CName', 'Mno'])
category_table = category_table[category_table.Cno < 12]
category_table
```

```python

measurement_table = pd.read_csv('beton24_csvs.csv/hamkorqurilish_main_table_olchov.csv', header=None, usecols=[0, 1], names
                            =['Mno', 'Mname'])
measurement_table.head(5)
```

```python
product_table.merge(category_table, on="Cno", how='inner')\
            .merge(measurement_table, on="Mno", how='inner')\
            .drop(columns=['Mname_y', 'Cno', 'Mno'])\
            .rename(columns={'Mname_x': 'Mname'})
```
