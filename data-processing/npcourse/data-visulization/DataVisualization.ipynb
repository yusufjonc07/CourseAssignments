```python
import statsmodels.api as sm
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Horizonal and vertical bar plot

```python
df = pd.read_csv("netflix_titles.csv")

# Vertical bar plot: Top 10 countries
country_counts = df['country'].value_counts().head(10)
country_counts.plot(kind='bar', color='tomato')
plt.title('Top 10 Countries on Netflix')
plt.xlabel('Country')
plt.ylabel('Number of Titles')
plt.show()

# Horizontal bar plot: Top 10 genres
genre_counts = df['listed_in'].str.split(',').explode().str.strip().value_counts().head(10)
genre_counts.plot(kind='barh', color='skyblue')
plt.title('Top 10 Genres on Netflix')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')
plt.show()
```

## Histogram of tip percentages

```python
tips = pd.read_csv('tips.csv')

# Calculate tip percentage
tips['tip_pct'] = tips['tip'] / tips['total_bill'] * 100

# Plot histogram
plt.figure(figsize=(8,5))
plt.hist(tips['tip_pct'], bins=20, color='skyblue', edgecolor='black')

# Labels and title
plt.title("Distribution of Tip Percentages", fontsize=14)
plt.xlabel("Tip Percentage (%)", fontsize=12)
plt.ylabel("Number of Customers", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
tips = pd.read_csv('tips.csv')

# Calculating tip percentage
tips['tip_pct'] = tips['tip'] / tips['total_bill'] * 100

# Plot density (KDE)
plt.figure(figsize=(8,5))
tips['tip_pct'].plot(kind='kde', color='orange', linewidth=2)

# Labels and title
plt.title("Density Plot of Tip Percentages", fontsize=14)
plt.xlabel("Tip Percentage (%)", fontsize=12)
plt.ylabel("Density", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# we have to show plot
plt.show()
```

## A seaborn regression/scatter plot

```python
tips = pd.read_csv('tips.csv')

# Basic regression (scatter) plot
plt.figure(figsize=(8,5))
sns.regplot(x='total_bill', y='tip', data=tips, color='teal', line_kws={"color": "red", "lw": 2})

# Labels and title
plt.title("Relationship Between Total Bill and Tip Amount", fontsize=14)
plt.xlabel("Total Bill ($)", fontsize=12)
plt.ylabel("Tip ($)", fontsize=12)
plt.grid(alpha=0.3)

plt.show()
```

## Pair plot matrix of statsmodels macro data

```python
import statsmodels.api as sm
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# Load the macroeconomic dataset
data = sm.datasets.macrodata.load_pandas().data

# Select relevant columns for the pair plot
columns = ['realgdp','realint', 'cpi', 'unemp']
data = data[columns]

# Create the pair plot
sns.pairplot(data, kind='scatter')
plt.suptitle("Pair Plot Matrix of Macroeconomic Variables", y=1.02)
plt.show()
```

## Box plot of tipping percentage by day

```python
import pandas as pd
import seaborn as sns


tips = pd.read_csv('tips.csv')

# Calculate tip percentage
tips['tip_pct'] = tips['tip'] / tips['total_bill'] * 100

box_plot = sns.catplot(
    x="tip_pct",
    y="day",
    kind="bar",
    data=tips,
    height=4,
    aspect=1.2
)
box_plot.fig.suptitle("Distribution of Tip Percentage by Day", fontsize=12)

plt.show()
```
