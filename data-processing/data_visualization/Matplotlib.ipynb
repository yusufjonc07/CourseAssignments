```python

import matplotlib.pyplot as plt
import numpy as np

plt.rc("figure", figsize=(10, 10))

plt.style.use('_mpl-gallery-nogrid')

# make data: correlated + noise
np.random.seed(1)
x = np.random.randn(5000)
y = 1.2 * x + np.random.randn(5000) / 3

# plot:
fig, ax = plt.subplots()

ax.hexbin(x, y, gridsize=20)

ax.set(xlim=(-2, 2), ylim=(-3, 3))

plt.show()
```

```python
fig, axes = plt.subplots(2, 2, sharex=True, sharey=True)
for i in range(2):
 for j in range(2):
     axes[i, j].hist(np.random.standard_normal(500), bins=50,
     color="#34ce23", alpha=0.5)
fig.subplots_adjust(left=None, bottom=None, right=None, top=None,
 wspace=0, hspace=0)
```

```python
from datetime import datetime
fig, ax = plt.subplots(figsize=(10, 7))
data = pd.read_csv("stock_details_5_years.csv", index_col="Date")

data.index = pd.to_datetime(data.index, utc=True).tz_convert(None)

apple_data = data[data.Company=='NVDA']
spx = apple_data["Close"]

spx.plot(ax=ax, color="#de3364")

crisis_data = [
 (datetime(2020, 3, 16), "COVID Market Crash (“Black Monday II”)"),
 (datetime(2022, 1, 3), "S&P 500 All-Time High Before the Bear Market"),
 (datetime(2023, 5, 24), "NVIDIA’s AI Earnings Explosion")
]
for date, label in crisis_data:
    ax.annotate(label, 
                xy=(date, spx.asof(date)+20),
                xytext=(date, spx.asof(date)+100),
                arrowprops=dict(facecolor="black", headwidth=4, width=2, headlength=4),
                horizontalalignment="right", 
                verticalalignment="top")
    
# Zoom in on 2007-2010
ax.set_xlim(["1/1/2019", "1/1/2024"])
# ax.set_ylim([2000, 7000])
ax.set_title("Important dates in the 2019-2024 Nvidia's stock price history")
```

### 9.2 Plotting with pandas and seaborn

#### Bar Plots

```python
df = pd.DataFrame(np.random.uniform(size=(6, 4)),
index=["one", "two", "three", "four", "five", "six"],
columns=pd.Index(["A", "B", "C", "D"], name="Genus"))
df.plot.bar(figsize=(10, 5),  alpha=0.7)
df.plot.barh(stacked=True, alpha=0.5, figsize=(6, 4))
```

#### pandas.crosstab - a convenient way to compute a simple frequency table from two DataFrame columns:

```python
tips = pd.read_csv('tips.csv')
tips.head()
pay_counts = pd.crosstab(tips['day'], tips['size'])
pay_counts = pay_counts.reindex(index=["Thur", "Fri", "Sat", "Sun"])
pay_counts.plot.bar(stacked=True, figsize=(7, 4))
```

```python
import seaborn as sns
sns.set(rc={'figure.figsize': (10, 6)})

tips["tip_pct"] = tips["tip"] / (tips["total_bill"] - tips["tip"])
tips.head()
sns.barplot(x="tip_pct", y="day", data=tips, orient="h")
```

```python
# sns.set_palette("Greys_r") ## grayscale background color for boxes
# sns.set_style("whitegrid") ## backgroud of whole area with grid
sns.barplot(x="tip_pct", y="day", hue="time", data=tips, orient="h")
```

### Scatter or Point Plots
Point plots or scatter plots can be a useful way of examining the relationship between
two one-dimensional data series.

```python
import statsmodels.api as sm
macro = sm.datasets.macrodata.load_pandas()
data = macro.data[["cpi", "m1", "tbilrate", "unemp"]]
trans_data = np.log(data).diff().dropna()
trans_data.tail()
```

```python
ax = sns.regplot(x="m1", y="unemp", data=trans_data)
```

```python
sns.pairplot(trans_data, diag_kind="kde", plot_kws={"alpha": 0.2})
```

### Facet Grids and Categorical Data
a categorical column as a col paramter to catplot method makes up the number of plots which will be generated

```python
sns.catplot(x="day", y="tip_pct", hue="time", row='time', col="smoker", kind="bar", data=tips[tips.tip_pct < 1])
```
